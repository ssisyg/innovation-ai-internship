```python
def scrape_finviz_news(ticker):
    url = f"https://finviz.com/quote.ashx?t={ticker}"
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/120.0.0.0 Safari/537.36",
        "Accept-Language": "en-US,en;q=0.9",
        "Referer": "https://finviz.com/"
    }
    
    try:
        response = requests.get(url, headers=headers, timeout=15)
        response.raise_for_status()
        soup = BeautifulSoup(response.text, "html.parser")
        
        news_table = soup.find("table", {"id": "news-table"})
        if not news_table:
            print(f"{ticker}: 未找到新闻表格")
            return []
        
        rows = news_table.find_all("tr")
        news_list = []
        current_date = None
        
        for row in rows:
            cols = row.find_all("td")
            if len(cols) < 2:
                continue
            
            date_col = cols[0].text.strip()
            headline = cols[1].a.text.strip() if cols[1].a else ""
            link = cols[1].a["href"] if cols[1].a else ""
            
            # Finviz 日期格式：有些行只有时间没有日期
            if len(date_col) > 8:
                current_date = date_col.split(" ")[0]
                time_str = date_col.split(" ")[1] if len(date_col.split(" ")) > 1 else ""
            else:
                time_str = date_col
            
            if headline and current_date:
                news_list.append({
                    "ticker": ticker,
                    "date": current_date,
                    "time": time_str,
                    "headline": headline,
                    "url": link,
                    "source": "finviz"
                })
        
        return news_list
    
    except Exception as e:
        print(f"{ticker} 抓取失败: {e}")
        return []
```

```python
all_news = []

for ticker in tickers:
    print(f"抓取 {ticker} 新闻...")
    news = scrape_finviz_news(ticker)
    all_news.extend(news)
    print(f"  获取 {len(news)} 条")
    time.sleep(3)  # 避免被封

df_news = pd.DataFrame(all_news)
print(f"\n总计抓取: {len(df_news)} 条新闻")
df_news.head(10)
```

```python
# 去重
df_news = df_news.drop_duplicates(subset=["ticker", "headline"])

# 过滤空标题
df_news = df_news[df_news["headline"].str.len() > 10]

# 统计每只股票新闻数
print("各股票新闻数量：")
print(df_news["ticker"].value_counts())
print(f"\n总计: {len(df_news)} 条")
```
