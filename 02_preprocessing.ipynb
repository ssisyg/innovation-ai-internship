```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
import os

tickers = [
    "AAPL", "MSFT", "GOOGL", "META", "AMZN", "NFLX", "TSLA", "ABNB",
    "NVDA", "AMD", "INTC", "QCOM", "AVGO", "MU", "ARM", "AMAT",
    "V", "MA", "PYPL", "COIN", "HOOD",
    "JPM", "GS", "BAC", "WFC", "MS", "BLK",
    "CRM", "NOW", "SNOW", "PLTR", "NET", "DDOG", "MDB",
    "WMT", "TGT", "COST", "EBAY", "SHOP",
    "UNH", "ISRG", "DXCM",
    "ENPH", "FSLR", "RIVN"
]
os.makedirs("data/processed", exist_ok=True)
```

```python
for ticker in tickers:
    df = pd.read_csv(f"data/raw/{ticker}_price.csv", index_col=0)
    df.index = pd.to_datetime(df.index)
    missing = df.isnull().sum().sum()
    print(f"{ticker}: {len(df)} 行, 缺失值 {missing} 个")
```

```python
from sklearn.preprocessing import MinMaxScaler

for ticker in tickers:
    df = pd.read_csv(f"data/raw/{ticker}_price.csv", index_col=0)
    df.index = pd.to_datetime(df.index)

    df = df.dropna()
    df = df.sort_index()

    scaler = MinMaxScaler()
    df_scaled = pd.DataFrame(
        scaler.fit_transform(df),
        index=df.index,
        columns=df.columns
    )

    df_scaled.to_csv(f"data/processed/{ticker}_clean.csv")
    print(f"{ticker} 处理完成，{len(df_scaled)} 行")
```

```python
print("=== 数据质量报告 ===")
for ticker in tickers:
    raw = pd.read_csv(f"data/raw/{ticker}_price.csv", header=[0,1], index_col=0)
    clean = pd.read_csv(f"data/processed/{ticker}_clean.csv", index_col=0)
    print(f"{ticker}: 原始 {len(raw)} 行 → 清洗后 {len(clean)} 行")
```
